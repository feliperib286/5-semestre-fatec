```python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# 1. Defina o seu RA da FATEC para a semente aleatória
SEU_RA = 42 # <-- SUBSTITUA AQUI

# 2. Gerar o conjunto de dados "Moons" (10.000 amostras, 15% de ruído)
X, y = make_moons(n_samples=10000, noise=0.15, random_state=SEU_RA)

# 3. Dividir o conjunto em treinamento (80%) e teste (20%)
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.2, random_state=SEU_RA)

# ---------------------------------------------------------
# 4. Treinamento e Avaliação do Classificador SVM
# ---------------------------------------------------------
print("Treinando classificador SVM...")
# O kernel 'rbf' é o mais adequado para separar dados não-lineares como o Moons
svm_clf = SVC(kernel='rbf', random_state=SEU_RA)
svm_clf.fit(X_treino, y_treino)

pred_svm = svm_clf.predict(X_teste)
acuracia_svm = accuracy_score(y_teste, pred_svm)
print(f"Acurácia do SVM no conjunto de teste: {acuracia_svm:.4f}\n")

# ---------------------------------------------------------
# 5. Treinamento da Árvore de Decisão com GridSearchCV
# ---------------------------------------------------------
print("Iniciando GridSearchCV para Árvore de Decisão...")

# Definindo a grade de hiperparâmetros a serem testados
parametros_busca = {
    'max_depth': [2, 3, 5, 10, None], # Profundidade máxima da árvore
    'min_samples_split': [2, 5, 10],  # Mínimo de amostras para dividir um nó
    'min_samples_leaf': [1, 2, 4]     # Mínimo de amostras que uma folha deve ter
}

arvore_clf = DecisionTreeClassifier(random_state=SEU_RA)

# Configurando a busca: cv=5 significa Validação Cruzada em 5 partes
grid_search = GridSearchCV(arvore_clf, parametros_busca, cv=5, scoring='accuracy')
grid_search.fit(X_treino, y_treino)

print(f"Melhores hiperparâmetros encontrados: {grid_search.best_params_}")

# Avaliando o melhor modelo encontrado pelo GridSearch no conjunto de teste
melhor_arvore = grid_search.best_estimator_
pred_arvore = melhor_arvore.predict(X_teste)
acuracia_arvore = accuracy_score(y_teste, pred_arvore)
print(f"Acurácia da melhor Árvore de Decisão: {acuracia_arvore:.4f}")
```
